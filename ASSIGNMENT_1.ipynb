```python
marks = [
    [80, 75, 90],
    [60, 70, 65],
    [95, 92, 88]
]

highest = 0
student = 0

for i in range(len(marks)):
    avg = sum(marks[i]) / len(marks[i])
    print("Student", i+1, "Average =", avg)

    if avg > highest:
        highest = avg
        student = i + 1

print("Highest Average =", highest)
print("Top Student =", student)
```

DICTIONARY

```python
student = {
    "Name": "Rahul",
    "Roll": 101,
    "Department": "CSE",
    "CGPA": 8.2
}

print(student)

student["CGPA"] = 8.8

print("Updated Dictionary:")
print(student)
```

```python
marks = {
    "Math": 90,
    "Physics": 85,
    "Chemistry": 95,
    "English": 88
}

print("Keys:", marks.keys())
print("Values:", marks.values())

highest = max(marks, key=marks.get)

print("Highest Marks Subject:", highest)
print("Marks:", marks[highest])
```

Dictionary in List

```python
students = [
    {"Name": "Rahul", "Marks": 80},
    {"Name": "Amit", "Marks": 92},
    {"Name": "Priya", "Marks": 88}
]

print("Student Names:")
for s in students:
    print(s["Name"])

topper = max(students, key=lambda x: x["Marks"])

print("Topper =", topper["Name"])
print("Marks =", topper["Marks"])
```

```python
books = [
    {"Title": "Python", "Price": 450},
    {"Title": "Java", "Price": 600},
    {"Title": "C++", "Price": 500}
]

expensive = max(books, key=lambda x: x["Price"])

print("Most Expensive Book:", expensive["Title"])

total = 0
for b in books:
    total += b["Price"]

print("Total Cost =", total)
```

ARRAY

```python
from array import *

arr = array('i', [12, 45, 23, 67, 34, 89, 21, 56, 78, 10])

print("Maximum =", max(arr))
print("Minimum =", min(arr))
print("Average =", sum(arr)/len(arr))
```

```python
from array import *

arr = array('i', [10, 20, 30, 40, 50])

arr.insert(2, 25)
print(arr)

arr.remove(40)
print(arr)

print("Reverse Order:")
for i in range(len(arr)-1, -1, -1):
    print(arr[i], end=" ")
```

SET

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union =", A | B)
print("Intersection =", A & B)
print("Difference =", A - B)
print("Symmetric Difference =", A ^ B)
```
